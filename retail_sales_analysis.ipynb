```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")
```

```python
from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(file_name, encoding='latin1')
```

```python
print("Dataset shape:", df.shape)

display(df.head())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe(include="all").T)
```

```python
# Create a working copy
df_clean = df.copy()

# Remove unnecessary whitespace from column names and text
df_clean.columns = df_clean.columns.str.strip()

for col in df_clean.select_dtypes(include="object").columns:
    df_clean[col] = df_clean[col].str.strip()

# Convert date columns to datetime
df_clean["Order Date"] = pd.to_datetime(
    df_clean["Order Date"], format="%m/%d/%Y"
)

df_clean["Ship Date"] = pd.to_datetime(
    df_clean["Ship Date"], format="%m/%d/%Y"
)

# Convert Postal Code to string
df_clean["Postal Code"] = df_clean["Postal Code"].astype("Int64").astype("string")

# Remove unnecessary Row ID
df_clean.drop(columns=["Row ID"], inplace=True)

# Handle missing postal codes
df_clean["Postal Code"] = df_clean["Postal Code"].fillna("Unknown")

# Create useful time-based features
df_clean["Order Year"] = df_clean["Order Date"].dt.year
df_clean["Order Month"] = df_clean["Order Date"].dt.month
df_clean["Order Month Name"] = df_clean["Order Date"].dt.month_name()
df_clean["Year-Month"] = df_clean["Order Date"].dt.to_period("M").astype(str)

# Calculate shipping duration
df_clean["Shipping Days"] = (
    df_clean["Ship Date"] - df_clean["Order Date"]
).dt.days

# Check results
print("Cleaned dataset shape:", df_clean.shape)
print("Duplicate records:", df_clean.duplicated().sum())

print("\nMissing values:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])

print("\nData types:")
print(df_clean.dtypes)

display(df_clean.head())
```

```python
total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_orders = df_clean["Order ID"].nunique()
total_customers = df_clean["Customer ID"].nunique()
total_products = df_clean["Product ID"].nunique()
average_order_value = total_sales / total_orders
profit_margin = (total_profit / total_sales) * 100

business_summary = pd.DataFrame({
    "Metric": [
        "Total Revenue",
        "Total Profit",
        "Total Orders",
        "Total Customers",
        "Unique Products",
        "Average Order Value",
        "Profit Margin"
    ],
    "Value": [
        total_sales,
        total_profit,
        total_orders,
        total_customers,
        total_products,
        average_order_value,
        profit_margin
    ]
})

display(business_summary)
```

```python
print("Negative sales:", (df_clean["Sales"] < 0).sum())
print("Negative profit records:", (df_clean["Profit"] < 0).sum())
print("Zero-profit records:", (df_clean["Profit"] == 0).sum())

print("\nDiscount distribution:")
display(df_clean["Discount"].describe())

print("\nShipping duration:")
display(df_clean["Shipping Days"].describe())
```

```python
# Calculate business KPIs
total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_orders = df_clean["Order ID"].nunique()
total_customers = df_clean["Customer ID"].nunique()

average_order_value = total_sales / total_orders
profit_margin = (total_profit / total_sales) * 100

print(f"Total Revenue: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
print(f"Average Order Value: ${average_order_value:,.2f}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

```python
monthly_sales = (
    df_clean.groupby("Year-Month")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
    .reset_index()
)

plt.figure(figsize=(15, 6))

sns.lineplot(
    data=monthly_sales,
    x="Year-Month",
    y="Sales",
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

```python
category_performance = (
    df_clean.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Quantity_Sold=("Quantity", "sum"),
        Orders=("Order ID", "nunique")
    )
    .reset_index()
)

category_performance["Profit_Margin"] = (
    category_performance["Total_Profit"] /
    category_performance["Total_Sales"] * 100
)

display(category_performance.sort_values("Total_Sales", ascending=False).round(2))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(
    data=category_performance,
    x="Category",
    y="Total_Sales",
    ax=axes[0]
)
axes[0].set_title("Revenue by Product Category")
axes[0].tick_params(axis="x", rotation=20)

sns.barplot(
    data=category_performance,
    x="Category",
    y="Total_Profit",
    ax=axes[1]
)
axes[1].set_title("Profit by Product Category")
axes[1].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.show()
```

```python
region_performance = (
    df_clean.groupby("Region")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Customers=("Customer ID", "nunique")
    )
    .reset_index()
)

region_performance["Profit_Margin"] = (
    region_performance["Total_Profit"] /
    region_performance["Total_Sales"] * 100
)

display(region_performance.sort_values("Total_Sales", ascending=False).round(2))
```

```python
segment_performance = (
    df_clean.groupby("Segment")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Customers=("Customer ID", "nunique"),
        Quantity_Sold=("Quantity", "sum")
    )
    .reset_index()
)

segment_performance["Profit_Margin"] = (
    segment_performance["Total_Profit"] /
    segment_performance["Total_Sales"] * 100
).round(2)

segment_performance.sort_values(
    "Total_Sales", ascending=False
)
```

```python
customer_performance = (
    df_clean.groupby(["Customer ID", "Customer Name"])
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity_Sold=("Quantity", "sum")
    )
    .reset_index()
)

customer_performance = customer_performance.sort_values(
    "Total_Sales", ascending=False
)

customer_performance.head(10).round(2)
```

```python
loss_making_customers = (
    customer_performance[
        customer_performance["Total_Profit"] < 0
    ]
    .sort_values("Total_Profit")
)

loss_making_customers.head(10).round(2)
```

```python
loss_customer_ids = loss_making_customers["Customer ID"].head(10)

loss_customer_analysis = (
    df_clean[df_clean["Customer ID"].isin(loss_customer_ids)]
    .groupby(["Customer ID", "Customer Name"])
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Average_Discount=("Discount", "mean"),
        Maximum_Discount=("Discount", "max"),
        Orders=("Order ID", "nunique")
    )
    .reset_index()
    .sort_values("Total_Profit")
)

loss_customer_analysis.round(2)
```
